```python
# Read the file with only the specified columns in this case the final gen (355 columns)
allele_freq = pd.read_csv(path_ldp_af, sep = '\t', usecols=final_gen)
```

```python
####calculating mean ########## 
## calculating the mean across subp insde optima
#allele_freq = allele_freq.groupby(allele_freq.columns.str.split('_').str[0], axis=1).mean()
##################

n_components = len(allele_freq.columns)

scaler = StandardScaler()
scaler.fit(allele_freq)
scaled = scaler.fit_transform(allele_freq)

scaled_allele_freq = pd.DataFrame(scaled, columns=allele_freq.columns)

# Perform PCA
pca = PCA(n_components=n_components)  # Set the desired number of components
pca.fit(scaled_allele_freq)

# Get the transformed data (projected onto the principal components)
transformed_data = pca.transform(scaled_allele_freq)

explain_var_ratio = pca.explained_variance_ratio_

cumulative_variance = np.cumsum(explain_var_ratio)

# Set the desired cumulative explained variance threshold
threshold = 0.96

# Find the number of components that exceed the threshold
num_components = np.sum(cumulative_variance <= threshold) + 1
```

```python
num_components
```

```python
# Open the file in write mode and write the integer
with open('LEA_analyses/num_components.txt', 'w') as file:
    file.write(str(num_components))
```

```python
# Open the file in read mode and read the integer
with open('LEA_analyses/num_components.txt', 'r') as file:
    number_from_file = file.read()
    number_as_int = int(number_from_file)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Assuming 'p_values' is your DataFrame containing p-values
# Assuming 'env' is your environmental variable DataFrame
alpha_adj = 0.05/13986
# Extract p-values corresponding to the first environmental variable
p_values_env1 = p_values.iloc[19, :]

# Create a data frame for plotting
manhattan_df = pd.DataFrame({
    'SNP': range(1, 13986),  # Assuming SNP IDs are row numbers
    'P': p_values_env1,                 # Assuming the first column of 'env' contains the first environmental variable
})

# Create Manhattan plot
plt.figure(figsize=(10, 6))
plt.scatter(manhattan_df['SNP'], -np.log10(manhattan_df['P']), c='blue', marker='o', alpha=0.5)
plt.axhline(y=-np.log10(alpha_adj), color='red', linestyle='--', label=f'Bonferroni threshold ({alpha_adj:.5f})')

plt.xlabel('SNP')
plt.ylabel('-log10(P-value)')
plt.title('Manhattan Plot for Environmental Variable 1')
plt.show()

```
