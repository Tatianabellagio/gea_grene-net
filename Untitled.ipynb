```r
# Load necessary library
library(topr)

# Load the data
data <- read.csv("path_to_your_data/gwas_region_data.csv")

# Plot the region of interest
regionplot(data = data,
           chr = "chromosome",      # specify the chromosome column
           pos = "position",        # specify the position column
           pval = "log_p",          # specify the log-transformed p-value column
           chr_interest = 2,        # specify the chromosome of interest
           start_pos = 9718274,     # start position of the region
           end_pos = 9888341,       # end position of the region
           title = "Region Plot for Chromosome 2: 9718274-9888341")

# Save the plot or display it
ggsave("region_plot.png")
```
