```python
import requests

def fetch_go_terms(gene_id):
    url = f"https://api.geneontology.org/api/bioentity/gene/{gene_id}/function"
    response = requests.get(url)
    
    if response.status_code == 200:
        data = response.json()
        for annotation in data['associations']:
            go_term = annotation['object']['id']
            go_name = annotation['object']['label']
            print(f"GO Term: {go_term}, Name: {go_name}")
    else:
        print(f"Error: {response.status_code}")
```

```python
import requests
import json

def perform_go_enrichment(gene_list):
    url = "https://api.geneontology.org/api/ontology/enrichment"
    
    # Organize data for the POST request
    data = {
        "gene_list": gene_list,
        "species": "human",  # Replace with your species of interest (e.g., arabidopsis)
        "aspects": ["biological_process"]  # Can be biological_process, molecular_function, or cellular_component
    }
    
    headers = {
        "Content-Type": "application/json"
    }
    
    # Make the request to GO API
    response = requests.post(url, data=json.dumps(data), headers=headers)
    
    if response.status_code == 200:
        enriched_terms = response.json()['results']
        for result in enriched_terms:
            print(f"GO Term: {result['term']['id']}, Name: {result['term']['label']}, P-value: {result['pValue']}")
    else:
        print(f"Error: {response.status_code}")

# Example gene list (replace with your gene IDs)
perform_go_enrichment('CAM5')
```

```python
import requests

def fetch_go_terms(gene_id):
    url = f"https://api.geneontology.org/api/bioentity/gene/TAIR:{gene_id}/function"
    response = requests.get(url)
    
    if response.status_code == 200:
        data = response.json()
        if 'associations' in data:
            for annotation in data['associations']:
                go_term = annotation['object']['id']
                go_name = annotation['object']['label']
                print(f"GO Term: {go_term}, Name: {go_name}")
        else:
            print(f"No GO terms found for {gene_id}")
    else:
        print(f"Error: {response.status_code}")

# Example with Arabidopsis gene IDs
arabidopsis_genes = ['CAM5']
for gene in arabidopsis_genes:
    fetch_go_terms(gene)
```

```python
import requests

def map_tair_to_uniprot(tair_ids):
    url = "https://www.uniprot.org/uploadlists/"
    params = {
        'from': 'TAIR_ID',
        'to': 'ACC',
        'format': 'tab',
        'query': ' '.join(tair_ids)
    }
    response = requests.get(url, params=params)
    if response.status_code == 200:
        print(response.text)
    else:
        print(f"Error: {response.status_code}")

tair_ids = ['AT1G26680', 'AT1G30720', 'AT1G30710', 'AT1G55660']
map_tair_to_uniprot(tair_ids)
```
