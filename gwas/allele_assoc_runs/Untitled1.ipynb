```r
# Base path
path <- "/carnegie/nobackup/scratch/tbellagio/gea_grene-net/idea_fromind_to_pop/allele_assoc_runs"

# Path for bslmm results, constructed with file.path for better handling of separators
path_bslmm <- file.path(path, "bslmm")

# Variable for the biological variable of interest

fdr_cut = function(PIP_vec, alpha = 0.1){
  p = sort(1 - PIP_vec)
  thres = cumsum(p)/seq_len(length(p))
  k = which.max(thres >= alpha)
  cutoff = 1 - p[k-1]
  return(cutoff)
}
```

```r
biovar = 'bio1'
```

```r
file_path <- file.path(path_bslmm, paste0(biovar, "/output/", biovar, ".param.txt"))

# Reading the table with specified options
assoc <- read.table(file_path, sep = '\t', header = TRUE)

# Calculate the cutoff value
cutoff <- fdr_cut(assoc$gamma)

# Filter the gamma values that are greater than or equal to the cutoff
significant_gammas <- assoc$gamma[assoc$gamma >= cutoff]
print(significant_gammas)
# Define the path where you want to save the CSV file
output_csv_path <- file.path(path_bslmm, paste0(biovar, "/output/significant_gammas.csv"))

# Use write.csv to save the data, row.names=FALSE omits row numbers in the output file
write.csv(data.frame(significant_gammas), output_csv_path, row.names = FALSE)

# Optionally, print a message that the file has been saved
cat("Significant gammas for", biovar, "saved to:", output_csv_path, "\n")
```

```r
biovars <- list.files(path = path_bslmm)
```

```r
# Loop over each biological variable
for (biovar in biovars) {
  # Constructing the full path to the file
  file_path <- file.path(path_bslmm, paste0(biovar, "/output/", biovar, ".param.txt"))

  # Reading the table with specified options
  assoc <- read.table(file_path, sep = '\t', header = TRUE)

  # Calculate the cutoff value
  cutoff <- fdr_cut(assoc$gamma)

  # Filter the gamma values that are greater than or equal to the cutoff
  significant_gammas <- assoc$gamma[assoc$gamma >= cutoff]
  print(significant_gammas)
  # Define the path where you want to save the CSV file
  output_csv_path <- file.path(path_bslmm, paste0(biovar, "/output/significant_gammas.csv"))

  # Use write.csv to save the data, row.names=FALSE omits row numbers in the output file
  write.csv(data.frame(significant_gammas), output_csv_path, row.names = FALSE)

  # Optionally, print a message that the file has been saved
  cat("Significant gammas for", biovar, "saved to:", output_csv_path, "\n")
}
```
