```python
import numpy as np
import pandas as pd
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import dask.dataframe as dd
import os
import dask.dataframe as dd
```

```python
final_gen = pd.read_csv('final_gen.csv')['sample_name'].to_list()
final_gen.append('0')
```

```python
path_all_af_indexed = '/carnegie/nobackup/scratch/tbellagio/gea_grene-net/baypass_terminal/merged_hapFIRE_allele_frequency_indexed.csv'
```

```python
allele_freq = dd.read_csv(path_all_af_indexed, sep = ',', usecols=final_gen)
```

```python
allele_freq = allele_freq.compute() 
```

```python
n_components = len(allele_freq.columns)
```

```python
n_components = len(allele_freq.columns)

scaler = StandardScaler()
scaler.fit(allele_freq)
scaled = scaler.fit_transform(allele_freq)

scaled_allele_freq = pd.DataFrame(scaled, columns=allele_freq.columns)

# Perform PCA
pca = PCA(n_components=n_components)  # Set the desired number of components
pca.fit(scaled_allele_freq)

# Get the transformed data (projected onto the principal components)
transformed_data = pca.transform(scaled_allele_freq)

explain_var_ratio = pca.explained_variance_ratio_

cumulative_variance = np.cumsum(explain_var_ratio)

# Set the desired cumulative explained variance threshold
threshold = 0.96

# Find the number of components that exceed the threshold
num_components = np.sum(cumulative_variance <= threshold) + 1
```

```python
# Open the file in write mode and write the integer
with open('LEA_analyses/full/num_components.txt', 'w') as file:
    file.write(str(num_components))
```

```python
# Open the file in read mode and read the integer
with open('LEA_analyses/full/num_components.txt', 'r') as file:
    number_from_file = file.read()
    number_as_int = int(number_from_file)
```

```python
# Path to the partitions 
path_all_af = '/carnegie/nobackup/scratch/tbellagio/gea_grene-net/allele_freq_partitions'

# Directory setup for results and shell scripts
path_analysis = '/carnegie/nobackup/scratch/tbellagio/gea_grene-net/LEA_analyses'
results_dir = '/carnegie/nobackup/scratch/tbellagio/gea_grene-net/LEA_analyses/full'
sh_dir = '/carnegie/nobackup/scratch/tbellagio/gea_grene-net/LEA_analyses/shfiles'
os.makedirs(results_dir, exist_ok=True)
os.makedirs(sh_dir, exist_ok=True)

# Function to create and save bash scripts
def create_bash_script(i, input_file):
    output_file = os.path.join(results_dir, f"result_{input_file.split('.')[0]}.out")  # Assuming input files are like 'file.csv'
    print(input_file)
    print(output_file)
    cmd = f'Rscript run_lfmm.R {input_file} {output_file}'
    script_content = f'''#!/bin/bash
#SBATCH --job-name=chain_{i+1}
#SBATCH --nodes=1
#SBATCH --ntasks=1
#SBATCH --mem-per-cpu=100gb
#SBATCH --cpus-per-task=10
#SBATCH --output={sh_dir}/chain_{i+1}_%j.out
#SBATCH --mail-user=tbellagio@carnegiescience.edu
#SBATCH --mail-type=FAIL

module load python/3.11_conda
conda activate /home/tbellagio/miniforge3/envs/snakemake_pipeline
cd /carnegie/nobackup/scratch/tbellagio/gea_grene-net/LEA_analyses
{cmd}
'''
    with open(os.path.join(sh_dir, f'chain_{i+1}.sh'), 'w') as f:
        f.write(script_content)

# Read allele frequency files and create scripts
chunk_counter = 0
for file in sorted(os.listdir(path_all_af)):
    if file.endswith(".csv"):  # Make sure to process only CSV files
        create_bash_script(chunk_counter, file)
        chunk_counter += 1
```
