```python
import pandas as pd
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
```

```python
data = pd.read_csv('/carnegie/nobackup/scratch/tbellagio/grene/data/bioclimvars_experimental_sites_era5.csv')
```

```python

# Drop the unnecessary columns
data_cleaned = data.drop(columns=['site'])

# Standardize the data
scaler = StandardScaler()
data_scaled = scaler.fit_transform(data_cleaned)

# Initialize PCA, using enough components to explain at least 95% of the variance
pca = PCA(n_components=0.95)
principal_components = pca.fit_transform(data_scaled)

# Create a DataFrame for the principal components
pca_df = pd.DataFrame(data=principal_components, columns=[f'PC{i+1}' for i in range(principal_components.shape[1])])

# Display the new DataFrame with principal components and the explained variance ratio
explained_variance_ratio = pca.explained_variance_ratio_

## add the site info 
pca_df = pd.concat([data['site'] ,pca_df],axis=1)
```

```python
pca_df.to_csv('pca_climate_var_experimental_sites.csv')
```
