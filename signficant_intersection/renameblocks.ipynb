```python
import pandas as pd
```

```python
import pickle
dict_blocks = '../key_files/blocks_snpsid_dict.pkl'

with open(dict_blocks, 'rb') as file:
    dict_blocks = pickle.load(file)

reverse_mapping = {item: key for key, values in dict_blocks.items() for item in values}
```

```python
bh_genes = pd.read_csv('genes_info_BH_tair10.csv')
```

```python
dict_blocks['1_485'][-1]
```

```python

# Replace the values in the `block_id` column with the first and last elements of the corresponding lists from `dict_blocks`
bh_genes['block_id'] = bh_genes['block_id'].map(
    lambda x: [dict_blocks[x][0], dict_blocks[x][-1]] if x in dict_blocks else None
)
```

```python
# Function to format the block_id column as `chrom:start-end`
def format_block_id(block):
    if block and len(block) == 2:
        chrom, start = block[0].split('_')
        _, end = block[1].split('_')
        return f"{chrom}:{start}-{end}"
    return None

# Applying the formatting function to the block_id column
bh_genes['block_id'] = bh_genes['block_id'].map(format_block_id)
```

```python
bh_genes.to_csv('bh_genes_tair_w_chrom_id.csv',index=None)
```
