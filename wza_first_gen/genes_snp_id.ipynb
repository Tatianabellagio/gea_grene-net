```python


# Read the file
file_path = '../key_files/TAIR10_GFF3_genes.gff'
gff_df = pd.read_csv(file_path, sep='\t', comment='#', header=None)

# Assigning appropriate column names for the GFF3 file format
gff_df.columns = ['seqname', 'source', 'feature', 'start', 'end', 'score', 'strand', 'frame', 'attribute']
```

```python
gff_df = gff_df[gff_df['seqname'].isin(['Chr1', 'Chr2', 'Chr3', 'Chr4', 'Chr5'])]

gff_df['chrom'] = gff_df['seqname'].str.replace('Chr', '')
```

```python
gff_df = gff_df.reset_index()
```

```python
dict_snps = '../key_files/var_pos_grenenet.csv'
dict_snps = pd.read_csv(dict_snps)
```

```python
gff_df['gene_id'] = gff_df['attribute'].str.split(';').str[0].replace('ID=', '')
```

```python
# Split the 'id' column into 'chrom' and 'pos'
dict_snps[['chrom', 'pos']] = dict_snps['id'].str.split('_', expand=True)
dict_snps['pos'] = dict_snps['pos'].astype(int)  # Ensure position is an integer for comparison
```

```python
#dict_snps = dict_snps[dict_snps['maf05filter'].notna()].reset_index(drop=True)
```

```python
dict_snps['chrom'] = dict_snps['chrom'].astype(int)
```

```python
gff_df['chrom'] = gff_df['chrom'].astype(int)
```

```python
from tqdm import tqdm

# Initialize tqdm to add progress to pandas operations
tqdm.pandas()

# Create a function to assign a gene based on chrom and position
def assign_gene(row, genes_df):
    chrom = row['chrom']
    pos = row['pos']
    
    # Filter genes that match the chromosome and where the position is between the gene's start and end
    gene_match = genes_df[(genes_df['chrom'] == chrom) & (genes_df['start'] <= pos) & (genes_df['end'] >= pos)]
    
    # If a match is found, assign the gene's ID; otherwise, return NaN
    if not gene_match.empty:
        return gene_match.iloc[0]['gene_id']  # You can modify this to return specific attributes (e.g., 'ID')
    else:
        return None

# Apply the function with a progress bar
dict_snps['assigned_gene'] = dict_snps.progress_apply(assign_gene, axis=1, genes_df=gff_df)
```
